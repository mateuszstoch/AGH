## Zadanie 4



```python
fix = px.scatter(auto_df,x="horsepower",y="displacement",trendline="ols")

results = px.get_trendline_results(fix)
model = results.iloc[0]["px_fit_results"]
a = model.params[1]   
b = model.params[0]   
equation = f"horsepower = {a:.2f} · displacement + ({b:.2f})"

fix.add_annotation(
    x=auto_df["horsepower"].max()+10, 
    y=auto_df["displacement"].max()+60,    
    text=equation,
    showarrow=False,
    font=dict(size=14, color="black")
)

print(equation)

fix.show()
```

Używając metody najmniejszych kwadratów, wyznaczyłem linię trendu, która pozwala przybliżyć wartość mocy silnika w zależności od pojemności silnika:

$$
\text{horsepower} = 2,44 \cdot \text{displacement} - 60,42
$$
